```python
import pandas as pd 

tabela = pd.read_csv("07-dados-abertos-precos-glp.csv", sep=";", encoding="utf-8-sig")
display(tabela)
```

```python
df = tabela.dropna(axis=1)
display(df)
display(df.info())
```

```python
df["Valor de Venda"] = df["Valor de Venda"].str.replace(",",".").astype(float)
display(df)
```

```python
vendas_regiao = df[["Regiao - Sigla", "Valor de Venda"]].groupby("Regiao - Sigla").sum().reset_index()
display(vendas_regiao)
```

```python
vendas_estado = df[["Estado - Sigla", "Valor de Venda"]].groupby("Estado - Sigla").sum().reset_index()
display(vendas_estado)
```

```python
bandeiras = df["Bandeira"].value_counts().reset_index()
bandeiras.columns = ["Bandeira", "Quantidade"]
display(bandeiras)
```

```python
bandeiras["Bandeira"] = bandeiras["Bandeira"].str.strip().str.upper()
limite = bandeiras["Quantidade"].sum()*0.02
bandeiras.loc[bandeiras["Quantidade"] < limite, "Bandeira"] = "Outras"
bandeiras = bandeiras.groupby("Bandeira", as_index=False)["Quantidade"].sum()

grafico_bandeiras = px.pie(bandeiras, names="Bandeira", values="Quantidade", title="Vendas por Bandeira")
grafico_bandeiras.update_traces(textposition="outside", textinfo="percent+label")
grafico_bandeiras.show()
```
