```python
import pandas as pd

tabela = pd.read_csv("selemob.csv", encoding="latin-1", sep=";")
display(tabela)
```

```python
df = tabela.copy()

for col in ["Valor de Contrapartida (R$)", "Valor de Empréstimo (R$)", "Valor de Investimento (R$)"]:
    df[col] = df[col].astype(str).replace("R$", "", regex=False).str.replace(" ", "",regex=False).str.replace(".","", regex=False).str.replace(".", ".", regex=False)
    df[col] = pd.to_numeric(df[col], errors="coerce")
display(df.head())
```
