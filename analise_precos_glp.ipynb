```python
import pandas as pd 

tabela = pd.read_csv("07-dados-abertos-precos-glp.csv", sep=";", encoding="utf-8-sig")
display(tabela)
```

```python
df = tabela.dropna(axis=1)
display(df)
display(df.info())
```

```python
df["Valor de Venda"] = df["Valor de Venda"].str.replace(",",".").astype(float)
display(df)
```

```python
vendas_regiao = df[["Regiao - Sigla", "Valor de Venda"]].groupby("Regiao - Sigla").sum()
display(vendas_regiao)
```

```python
vendas_estado = df[["Estado - Sigla", "Valor de Venda"]].groupby("Estado - Sigla").sum()
display(vendas_estado)
```

```python
vendas_municipio = df[["Municipio", "Valor de Venda"]].groupby("Municipio").sum()
display(vendas_municipio)
```

```python
import plotly.express as px

grafico_regioes = px.bar(x=df["Regiao - Sigla"], y=df["Valor de Venda"], title="Vendas por regiao")
grafico_regioes.update_yaxes(title="Valor da venda") 
grafico_regioes.update_xaxes(title="Regiao")
grafico_regioes.update_traces(marker_color="#E53935")
grafico_regioes.show()
```

```python

grafico_estados = px.bar(x=df["Estado - Sigla"], y=df["Valor de Venda"], title="Vendas por estado")
grafico_estados.update_yaxes(title="Valor de venda")
grafico_estados.update_xaxes(title="Estado")
grafico_estados.update_traces(marker_color="#636EFA", opacity=1)
grafico_estados.show()
```
