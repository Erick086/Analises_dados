```python
import pandas as pd 

tabela = pd.read_csv("Equipe_Nebula.csv")
display(tabela)
```

```python
df = tabela.copy()
```

```python
df["total_vendas"] = df["quantidade"] * df["preco_unitario"]
df["custo_total"] = df["quantidade"] * df["custo"]
df["lucro"] = df["total_vendas"] - df["custo_total"]
df["margem_lucro"] = ((df["lucro"] / df["total_vendas"]) * 100).round(2)
display(df)
```

```python
# FATURAMENTO TOTAL DA EQUIPE NEBULA

faturamento_total = sum(df["total_vendas"])
print(f"{faturamento_total:,.2f}")
```

```python
# LUCRO TOTAL 

lucro_total = sum(df["lucro"])
print(f"{lucro_total:,.2f}")
```

```python
# VOLUME DE VENDAS POR VENDEDOR

analise_vendedores = df[["vendedor", "quantidade"]].groupby("vendedor").sum()
display(analise_vendedores.sort_values(by="quantidade", ascending=False))
```

```python
# RECEITA TOTAL DE CADA VENDEDOR

analise_receita = df[["vendedor", "total_vendas"]].groupby("vendedor").sum()
analise_receita["total_vendas"] = (analise_receita["total_vendas"]/1_000_000).round(2)
display(analise_receita.sort_values(by="total_vendas", ascending=False))

```

```python
# TABELA DE VENDAS DE SERVICOS POR QUANTIDADE

analise_servicos = df[["servico", "quantidade"]].groupby("servico").sum()
display(analise_servicos.sort_values(by="quantidade", ascending=False ))
```

```python
# TABELA DE SERVICOS POR RECEITA

analise_receita_srvç = df[["servico", "total_vendas"]].groupby("servico").sum()
analise_receita_srvç["total_vendas"] = (analise_receita_srvç["total_vendas"]/1_000_000).round(2)
display(analise_receita_srvç.sort_values(by="total_vendas", ascending=False))
```

```python
# DISTRIBUICAO DE VENDAS POR ESTADO

analise_regioes = df[["total_vendas", "estado"]].groupby("estado").sum()
analise_regioes["total_vendas"] = (analise_regioes["total_vendas"]/1_000_000).round(2)
display(analise_regioes.sort_values(by="total_vendas", ascending=False))
```

```python
# MARGEM DE LUCRO MEDIA POR SERVICO

margem_media_lucro = df[["servico","margem_lucro"]].groupby("servico").mean().round(2)
display(margem_media_lucro.sort_values(by="margem_lucro", ascending=False))
```

```python
# EVOLUCAO DA RECEITA POR ANO

df["data_venda"] = pd.to_datetime(df["data_venda"])

receita_anual = df[["data_venda", "total_vendas"]].groupby(pd.Grouper(key="data_venda", freq="YE")).sum().sort_index(ascending=True)
receita_anual["Receita R$ M"] = (receita_anual["total_vendas"]/1_000_000).round(2)
receita_anual["Variacao %"] = (receita_anual["Receita R$ M"].pct_change()*100).round(2)
display(receita_anual)
```

```python
# MARGEM DE LUCRO MEDIA DE CADA VENDEDOR

margem_lucro_vendedores = df[["vendedor", "margem_lucro"]].groupby("vendedor").mean().round(2)
display(margem_lucro_vendedores.sort_values(by="margem_lucro", ascending=False))
```

```python
# TICKET MEDIO POR SERVICO

ticket_medio_srvç = df[["servico", "total_vendas"]].groupby("servico").sum()
ticket_medio_srvç["total_vendas"] = (ticket_medio_srvç["total_vendas"]/1_000_000).round(2)
display(ticket_medio_srvç.sort_values(by="total_vendas", ascending=False))
```

```python
# RELACAO DE FATURAMENTO DA EMPRESA POR CLIENTE

fat_por_cliente = df[["cliente", "total_vendas"]].groupby("cliente").sum()
display(fat_por_cliente.sort_values(by="total_vendas", ascending=False))
```

```python
# RANK DE EFICIENCIA POR VENDEDOR 

ranking_eficiencia = df.groupby("vendedor", as_index=False).agg(lucro_total = ("lucro", "sum"), qtd_total = ("quantidade", "sum"))

ranking_eficiencia["rentabilidade_venda"] = ranking_eficiencia["lucro_total"]/ranking_eficiencia["qtd_total"]

ranking_eficiencia = ranking_eficiencia.sort_values(by="rentabilidade_venda", ascending=False).reset_index(drop=True)

ranking_eficiencia["lucro_total"] = (ranking_eficiencia["lucro_total"]/1_000_000).round(2)

ranking_eficiencia["rentabilidade_venda"] = ranking_eficiencia["rentabilidade_venda"].round(2)
ranking_eficiencia["lucro_total"] = ranking_eficiencia["lucro_total"].round(2)
display(ranking_eficiencia)
```

```python
# QUANTO CADA SERVICO REPRESENTA DO FATURAMENTO TOTAL DA EMPRESA

faturamento_servico = df.groupby("servico",  as_index=False)["total_vendas"].sum()
#faturamento_total = df["total_vendas"].sum()
faturamento_servico["percentual"] = (faturamento_servico["total_vendas"]/faturamento_total)*100
faturamento_servico = faturamento_servico.sort_values(by="percentual", ascending=False)
faturamento_servico["total_vendas"] = (faturamento_servico["total_vendas"]/1_000_000).round(2)
faturamento_servico["percentual"] = faturamento_servico["percentual"].round(2)

faturamento_servico["percentual_fmt"] = faturamento_servico["percentual"].astype(str)+"%"
display(faturamento_servico)
```

```python
# RELACAO DO LUCRO ENTRE REGIAO E CATEGORIA

lucro_cat_reg = df.groupby(["regiao", "categoria_servico"]).agg(Total_vendas = ("total_vendas", "sum"))
lucro_cat_reg["Total_vendas"] = (lucro_cat_reg["Total_vendas"]/1_000_000).round(2)
display(lucro_cat_reg)
```

```python
# iDENTIFICANDO O MOTIVO DA QUEDA DE 60% NO FATURAMENTO ANUAL DA EMLRESA EM 2025

df["ano"] = df["data_venda"].dt.year
df["mes"] = df["data_venda"].dt.month_name(locale="pt_BR.utf8")
df["mes_num"] = df["data_venda"].dt.month

df_2025 = df[df["ano"]==2025]
faturamento_mes = df_2025.groupby(["mes_num", "mes"], as_index=False).agg(Faturamento_total = ("total_vendas", "sum"))

display(faturamento_mes.sort_values(by="Faturamento_total", ascending=False))

# COMO PODEMOS ANALISAR NA TABELA ABAIXO, HOUVE REGISTRO DE FATURAMENTO SOMENTE NOS MESES DE JANEIRO, MARCO, ABRIL E MAIO. POR SE TRATAR DE UMA BASE DE DAODS SINTETICOS, NAO POSSO AFIRMAR SE A AUSENCIA DOS DEMAIS MESES SE DEU PELA FALTA DE FATURAMENTO NELES OU PELA FALTA DE REGISTRO
```
