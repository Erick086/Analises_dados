```python
import pandas as pd

df = pd.read_csv("vendas_autoveiculos_brasil_bcb.csv", sep=";")
display(df)
```

```python
df["data"] = pd.to_datetime(df["data"], format="%d/%m/%Y")

vendas_anos = df.groupby(df["data"].dt.year)["valor"].sum().reset_index()
display(vendas_anos)
```

```python
import plotly.express as px

grafico = px.bar(x=vendas_anos["data"], y=vendas_anos["valor"], title="Analise de vendas de autoveiculos")
grafico.update_yaxes(title = "Valores")
grafico.update_xaxes(title= "Anos")
grafico.show()
```
